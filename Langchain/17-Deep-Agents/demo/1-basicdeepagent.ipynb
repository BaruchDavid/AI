### Deep Agents overview

build agents that can plan, use subagents and leverage file systems for complex tasks

### When to use deep agents

Use deep agents when you need agents that can:

* Handle complex, multi-step tasks that require planning and decomposition
* Manage large amounts of context through file system tools
* Delegate work to specialized subagents for context isolation
* Persist memory across conversations and threads

```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["TAVILY_API_KEY"] = os.getenv("TAVILY_API_KEY")
```

```python
### Tools- Internet search tool using Tavily's API

from tavily import TavilyClient
from typing import Literal, Optional
tavily_client = TavilyClient(api_key=os.environ["TAVILY_API_KEY"])

def web_search(query:str, max_results:int=5, topic:Optional[Literal["news", "images", "videos", "all"]]=None, include_raw_content:bool=False):
    """Run a Web search using Tavily's API and return the results."""
    
    return tavily_client.search(query, max_results=max_results, topic=topic, include_raw_content=include_raw_content)
```

```python
### Create a deep agent that can use the web search tool
## Prompt

## agent
## diese Lib ist für ein normalen agenten
from langchain.agents import create_agent

## diese Lib ist für einen deep-agent
from deepagents import create_deep_agent
from langchain.chat_models import init_chat_model

groq_models =init_chat_model(model="groq:openai/gpt-oss-20b")

deep_agent = create_deep_agent(model = groq_models, 
                  tools=[web_search], 
                  system_prompt="You are a helpful assistant that can search the web for information using the Tavily API. Use the web_search tool to find relevant information when needed.",)

results = deep_agent.invoke({"messages": [{"role": "user", "content": "What is the latest news about AI?"}]})

for m in results["messages"]:
    print(type(m), getattr(m, "content", m))
```
